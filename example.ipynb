# Using the SigmundAI extension for Jupyter

First, install the extension. The extension is updated regularly, so make sure you upgrade to the latest version.

```python
pip install jupyter-extension-sigmund --upgrade
```

Next, load the extension.

```python
%load_ext jupyter_extension_sigmund
```

Once the extension is loaded, you can connect to Sigmund. Make sure that <https://sigmundai.eu> is open in a browser tab and that you are logged in. The extension will automatically connect.

```python
%sigmund_connect
```

Once connected, you can send single-line messages to Sigmund using the `%sigmund <message>` magic command. Sigmund is able to run code on your computer, and receives the result of the code execution.

```python
%sigmund "Can you draw a random number and tell me what it is?"
```

You can send multiline messages with the `%%sigmund` cell magic.

```python
%%sigmund
Can you draw a randomly colored square in a figure and tell me what color it is?
Don't print the color to the console as text. I want you to deduce it by inspecting the figure.
```

To disconnect, run `%sigmund_disconnect`.

```python
%sigmund_disconnect
```
